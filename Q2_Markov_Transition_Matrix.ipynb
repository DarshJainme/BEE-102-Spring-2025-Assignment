## Q2. Building Markov transition matrix
- **INPUT** : A nucleotide sequence
- **OUTPUT** : A Transition matrix in the form of a dictionary

```python

import pandas as pd
from collections import defaultdict

sequence = "CTTCATGTGAAAGCAGCGTAAGTCA"

# Initialize counts
transitions = defaultdict(lambda: defaultdict(int))

# Count transitions
for i in range(len(sequence) - 1):
    curr = sequence[i]
    next_ = sequence[i + 1]
    transitions[curr][next_] += 1

# Convert to DataFrame
nucleotides = ['A', 'C', 'G', 'T']
matrix = pd.DataFrame(index=nucleotides, columns=nucleotides, dtype=float)

# Fill matrix with probabilities
for curr in nucleotides:
    total = sum(transitions[curr].values())
    for next_ in nucleotides:
        matrix.loc[curr, next_] = transitions[curr][next_] / total if total > 0 else 0.0

# Display transition matrix
print("Markov Transition Matrix: is:")
print(matrix)
```
