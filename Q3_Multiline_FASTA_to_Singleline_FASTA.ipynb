#### **Q3. Converting a Multi-line sequence fasta file to single-line fasta file**

- Input : A FASTA file with multiline sequences. A header line (Starting with **'>'**) comes before the start of each sequence in the input file
- Output : A FASTA file with the multiline sequences converted to single line sequences. The format of the header line is conserved in the output file

```python
file_path='multiline_input.fasta'
file=open(file_path,'r')
# print(file.read())
initialstring=""
a=0
output_file=open('multiline_output.fasta','w')
for line in file:
    # print(line)
    line=line.strip() 
    if(line[0]=='>' ):
        if(a==0):
            output_file.write(f'>{a+1}\n')
            #as first time no newline must be there
        else:
            output_file.write(f'\n>{a+1}\n')
            #other times there must be newline 
        a+=1
            
    else:
        output_file.write(line)
# the last string is actually not done yet.
output_file.close()
file.close()

print("The Multi-line FASTA file has been formatted into a Single line FASTA file and saved as 'multiline_output.fasta'.")
```
